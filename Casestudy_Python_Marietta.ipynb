```python
employees=[{"id":101,"name":"Alice","department":"Data","salary":65000,"rating":4.5},
           {"id":102,"name":"Bob","department":"AI","salary":72000,"rating":4.8},
           {"id":103,"name":"Charlie","department":"Web","salary":58000,"rating":3.9},
           {"id":104,"name":"David","department":"Data","salary":61000,"rating":4.2},
           {"id":105,"name":"Eva","department":"AI","salary":75000,"rating":4.9},
           {"id":106,"name":"Frank","department":"Cloud","salary":68000,"rating":4.1},
           {"id":107,"name":"Grace","department":"Web","salary":59000,"rating":3.8},
           {"id":108,"name":"Helen","department":"Data","salary":70000,"rating":4.6},]
totalno=(len(employees))
print(f"Total Employees:{totalno}\n")
salaries=[emp["salary"] for emp in employees]
avg_salary=sum(salaries)/totalno
print(f"Average Salary:{avg_salary}")
print(f"Highest Salary:{max(salaries)}")
print(f"Lowest Salary:{min(salaries)}\n")
print(f"Employees with Salary >65000 and Rating >=4.5")
for emp in employees:
  if emp["salary"]>65000 and emp["rating"]>=4.5:
   print(emp["name"])
newemployees={emp["id"]:emp["name"] for emp in employees}
print(f"\nEmployee Dictionary:{newemployees}")
newlist=[emp["name"] for emp in employees]
newlist.sort(reverse=True)
print(f"\nFirst three employees:{newlist[:3]}")
newlist.reverse()
print(f"First three employees:{newlist[:3]}")
newid=tuple(emp["id"] for emp in employees)
print(f"\nmaximum Employee ID:{max(newid)}")
print(f"minimum Employee ID:{min(newid)}")
newdep={emp["department"] for emp in employees}
print(f"Departments:{newdep}")

try:
    search_id = int(input("Enter Employee ID: "))
    found_employee = None
    for emp in employees:
        if emp["id"] == search_id:
            found_employee = emp
            break

    if found_employee:
        print("\nEmployee Found")
        print(f"ID: {found_employee['id']}")
        print(f"Name: {found_employee['name']}")
        print(f"Department: {found_employee['department']}")
        print(f"Salary: {found_employee['salary']}")
        print(f"Rating: {found_employee['rating']}")
    else:
        print("\nEmployee Not Found")
except ValueError:
    print("\nInvalid input. Please enter a valid numerical Employee ID.")
```
